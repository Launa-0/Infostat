```python
weight=float(input("몸무게(kg): "))
height=float(input("키(m): "))
```

```python
bmi=weight/height**2
```

```python
print(f"키: {height*100},(cm)")
print(f"몸무게: {weight} (kg)")
print(f"---> BMI = {bmi:.2f}")
```

```python
#3
principal=1000000
rate=0.05
n=4
year=10
```

```python
amount=principal*(1+rate/n)**(n*year)
```

```python
print(f"예치금액 {principal:,} (원)")
print(f"년간 이자율: {rate*100}% 분기마다 이자 계산을 할 때 {year}년후 수령액: {amount:,.0f}원")
```

```python
#5
won=int(input("환전하고자 하는 원화 금액(원): "))
```

```python
u_rate=1509.1
c_rate=219.2
j_rate=9.46
e_rate=1740.8
```

```python
print("[환율 변화 금액]")
print(f"원화: {won:,}원")
print(f"달러: {won/u_rate:,.2f} USD")
print(f"유로: {won/e_rate:,.2f} EUR")
print(f"위엔: {won/c_rate:,.0f} CNY")
print(f"엔화: {won/j_rate:,.0f} JPY")
```
