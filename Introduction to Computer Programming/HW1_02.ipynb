20230808 컴퓨터학과 오승현

```python
birth = "050825"
```

```python
# 문자수
length = len(birth)
```

```python
# 문자열 자르기
year = birth[0:2]
month = birth[2:4]
day = birth[4:6]
```

```python
# int형으로 변환
year = int(year) + 2000
month = int(month)
day = int(day)
```

```python
# 출력
print("생년월일:", birth)
print("문자수:", length)
print("년도:", year)
print("월:", month)
print("일:", day)
```
