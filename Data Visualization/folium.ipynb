```python
import pandas as pd
```

```python
location = "https://data.smartdublin.ie/dataset/33ec9fe2-4957-4e9a-ab55-c5e917c7a9ab/resource/2dec86ed-76ed-47a3-ae28-646db5c5b965/download/dublin.csv"
bike_station_locations = pd.read_csv(location)
```

```python
print(bike_station_locations)
```

```python
bike_station_locations = bike_station_locations[["Latitude", "Longitude", "Name"]]
```

```python
print(bike_station_locations)
```

```python
map = folium.Map(location=[bike_station_locations.Latitude.mean(), bike_station_locations.Longitude.mean()], zoom_start=12, control_scale=True)
```

```python
map
```

```python
for index, location_info in bike_station_locations.iterrows():
    folium.Marker([location_info["Latitude"], location_info["Longitude"]], popup=location_info["Name"]).add_to(map)
```

```python
map
```

```python
m = folium.Map(location=[37.607363, 127.042933], zoom_start=12)

folium.Marker(
    location=[37.607363, 127.042933],
    popup="Mt. Hood Meadows",
    icon=folium.Icon(icon="cloud"),
).add_to(m)

folium.Marker(
    location=[37.5999, 127.031111],
    popup="Timberline Lodge",
    icon=folium.Icon(color="green"),
).add_to(m)

folium.Marker(
    location=[37.6000, 127.02301],
    popup="Some Other Location",
    icon=folium.Icon(color="red", icon="info-sign"),
).add_to(m)


m
```

```python
m = folium.Map(location=[37.607363, 127.042933], tiles="CartoDB positron", zoom_start=13)

folium.Circle(
    radius=100,
    location=[37.607363, 127.042933],
    popup="동덕여대대",
    color="crimson",
    fill=False,
).add_to(m)

m
```

```python
m = folium.Map(location=[37.607363, 127.042933], zoom_start=13)

folium.CircleMarker(
    location=[37.607363, 127.042933],
    radius=50,
    popup="동덕여대",
    color="#3186cc",
    fill=True,
    fill_color="#3186cc",
).add_to(m)

m
```

```python
m = folium.Map(location=[-71.38, -73.9], zoom_start=11)

trail_coordinates = [
    (-71.351871840295871, -73.655963711222626),
    (-71.374144382613707, -73.719861619751498),
    (-71.391042575973145, -73.784922248007007),
    (-71.400964450973134, -73.851042243124397),
    (-71.402411391077322, -74.050048183880477),
]


folium.PolyLine(trail_coordinates, tooltip="Coast").add_to(m)

m
```
